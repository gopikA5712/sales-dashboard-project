```python
import zipfile

with zipfile.ZipFile("sales-forecasting.zip", 'r') as zip_ref:
    zip_ref.extractall("sales_data")

print("✅ Dataset Extracted")
```

```python
import pandas as pd

df = pd.read_csv("sales_data/train.csv")
df.head()
```

```python
# Convert date
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Month'] = df['Order Date'].dt.to_period('M').astype(str)
```

```python
print(df.columns)
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
total_sales = df['Sales'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()

print("Total Sales:", total_sales)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
```

```python
import plotly.express as px

monthly_sales = df.groupby('Month')['Sales'].sum().reset_index()

fig1 = px.line(monthly_sales, x='Month', y='Sales',
               title="Monthly Sales Trend")

fig1.show()
```

```python
region_sales = df.groupby('Region')['Sales'].sum().reset_index()

fig2 = px.bar(region_sales, x='Region', y='Sales',
              title="Sales by Region")

fig2.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().reset_index()

fig3 = px.pie(category_sales, names='Category', values='Sales',
              title="Sales Distribution by Category")

fig3.show()
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().nlargest(10).reset_index()

fig4 = px.bar(top_products, x='Sales', y='Product Name',
              orientation='h',
              title="Top 10 Products by Sales")

fig4.show()
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum().reset_index()

fig5 = px.bar(segment_sales, x='Segment', y='Sales',
              title="Sales by Customer Segment")

fig5.show()
```
